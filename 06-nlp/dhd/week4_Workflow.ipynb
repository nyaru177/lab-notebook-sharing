# NLP 模型演进学习工作流：RNN → Transformer → BERT

- **作者**：邓涵丹
- **日期**：2026-07-27
- **适用周次**：Week 4
- **分类**：NLP / PyTorch / Transformers
- **关键词**：RNN、LSTM、Transformer、BERT、文本分类
- **适合读者**：具备 Python 基础、了解神经网络基本概念

## 学习目标

1. 掌握 RNN/LSTM 序列建模的基本思想，并手写 BiLSTM 文本分类器。
2. 理解 Transformer 的自注意力机制，并实现一个简化版 Transformer 编码器。
3. 学会使用 HuggingFace 对 BERT 进行微调，完成文本分类任务。

## 1. 背景与问题

自然语言处理（NLP）的核心任务之一是将文本转化为计算机可理解的向量表示，并用于分类、生成等下游任务。

本工作流沿着 NLP 模型演进的主线，从基础的 RNN 序列建模，到引入注意力机制的 Transformer，再到预训练语言模型 BERT，逐步深入。这些模型是不良信息检测、情感分析、文本分类等实际应用的基础。

## 2. 整体工作流

本工作流分为三个独立的 Notebook，每个 Notebook 聚焦一个模型，并逐步递进：

- **[week4_BiLSTM-Classifier.ipynb](./week4_BiLSTM-Classifier.ipynb)**：手写 BiLSTM 文本分类器，使用注意力池化，在 AG_NEWS 数据集上训练。
- **[week4_Transformer-FromScratch.ipynb](./week4_Transformer-FromScratch.ipynb)**：从零实现 Transformer 编码器，包含多头注意力和位置编码，用于文本分类。
- **[week4_BERT-Finetune.ipynb](./week4_BERT-Finetune.ipynb)**：使用 HuggingFace `transformers` 库，对 `bert-base-uncased` 进行微调，完成 IMDB 情感分类。

所有中间产物（模型权重、分词器、日志）均保存在 `./intermediate_data/` 目录下。

## 3. 环境与依赖

建议使用 Python 3.8+，并安装以下依赖（详见 `requirements.txt`）：

- `torch>=1.10.0`
- `transformers>=4.20.0`
- `datasets>=2.0.0`（用于加载 IMDB）
- `matplotlib`, `seaborn`, `scikit-learn`, `tqdm`
- `watermark`（用于记录环境信息）

运行前请安装：`pip install -r requirements.txt`

## 4. 数据来源与预处理说明

- **AG_NEWS**（用于 BiLSTM 和 Transformer）：包含 4 类新闻文本，将简化为二分类（将类别 1、2 归为正类，3、4 归为负类）或直接使用多分类。为简单，本例使用二分类。
- **IMDB**（用于 BERT 微调）：电影评论，二分类（正面/负面）。

所有数据集均通过 `datasets` 自动下载。
