# 一、数组基础

```python
import numpy as np

print("NumPy 版本：", np.__version__)

# 创建固定随机数生成器，使每次运行结果相同
rng = np.random.default_rng(42)
```

```python
a = np.array([1, 2, 3, 4, 5])
b = np.zeros((3, 4)) #生成3行4列的全零数组，形状是(3,4)
c = np.ones((2, 3), dtype=np.float32)
d = np.eye(4)
e = np.arange(0, 10, 2)  #从0开始，到10之前结束，步长为2
f = np.linspace(0, 1, 5)  #含义是把闭区间 [0, 1] 均匀分成5个点
g = rng.normal(size=(3, 4))

print("a =", a)
print("\nb =\n", b)
print("\nc =\n", c)
print("\nd =\n", d)
print("\ne =", e)
print("\nf =", f)
print("\ng =\n", g)
```

```python
print("形状 shape：", g.shape)
print("维度数 ndim：", g.ndim)
print("元素总数 size：", g.size)
print("数据类型 dtype：", g.dtype)
```

# 三、理解axis

```python
x = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("原数组：")
print(x)

print("\n全部求和：", x.sum())
print("axis=0：", x.sum(axis=0))  #消除第0维，也就是把各行压在一起，得到每列的结果
print("axis=1：", x.sum(axis=1))  #消除第1维，也就是把各列压在一起，得到每行的结果
```

# 四、理解keepdims=True

```python
data = rng.normal(size=(4, 3))

mean_without_keepdims = data.mean(axis=1)
mean_with_keepdims = data.mean(axis=1, keepdims=True)

print("data shape：", data.shape)
print("不使用 keepdims：", mean_without_keepdims.shape)
print("使用 keepdims：", mean_with_keepdims.shape) #keepdims=True 表示求均值后仍保留原来的维度位置
```

# 五、理解广播机制
广播判断规则：从最右侧维度开始比较。两个维度满足以下任意条件，就可以广播：1，两个维度相等；2，其中一个维度是1；3，某个数组缺少这一维，可以在左侧补1。

a = np.ones((4, 1, 3))
b = np.ones((5, 3))

c = a + b #给 b左侧补1 b：1 × 5 × 3   a：4 × 1 × 3

print(a.shape)
print(b.shape)
print(c.shape) #4 和 1：可以广播,1 和 5：可以广播,3 和 3：相同

## 1，z-score标准化
目标让每一列：均值约为0，标准差约为1.

```python
data = rng.normal(size=(100, 5))

# 请补充
mean = data.mean(axis=0, keepdims=True)
std = data.std(axis=0, keepdims=True)
data_normalized = (data - mean) / std

print(data_normalized.mean(axis=0))
print(data_normalized.std(axis=0))

assert data_normalized.shape == data.shape
assert np.allclose(data_normalized.mean(axis=0), 0)
assert np.allclose(data_normalized.std(axis=0), 1)
```

## 2,余弦相似度矩阵

```python
A = rng.normal(size=(10, 128)) #表示有10个128维向量。
assert A.shape == (10, 128)
norms = np.linalg.norm(A, axis=1, keepdims=True) #首先计算每个向量的长度
A_normalized = A / norms #然后归一化
similarities = A_normalized @ A_normalized.T #最后矩阵乘法
print(np.diag(similarities))
```

## 3,数值稳定版softmax

当输入很大时可能溢出，因此先减去每行最大值，再计算指数，检查每行概率和

```python
import numpy as np
def softmax(x):
    """对数组最后一个维度计算数值稳定的 softmax。"""
    x = np.asarray(x, dtype=np.float64)
    if x.ndim == 0:
        raise ValueError("x 至少应为一维数组")
    # 每组数据减去该组最大值，避免 exp 溢出
    shifted = x - np.max(x, axis=-1, keepdims=True)
    exp_x = np.exp(shifted)
    result = exp_x / exp_x.sum(axis=-1, keepdims=True)
    # 自检：输出形状不变
    assert result.shape == x.shape
    # 自检：最后一个维度的概率和为1
    assert np.allclose(result.sum(axis=-1), 1)
    return result
# 测试数据：包含较大的数值
x = np.array([
    [1000, 1001, 1002],
    [1, 2, 3],
    [-1000, -999, -998]
])
result = softmax(x)
print("Softmax 结果：")
print(result)
print("\n每行概率之和：")
print(result.sum(axis=-1))
```

## 4，one-hot编码

```python
labels = np.array([0, 2, 1])
num_classes=3
# 创建全零矩阵，形状为：样本数 × 类别数
result = np.zeros(
    (labels.size, num_classes),
    dtype=np.float32
)
result[np.arange(labels.size), labels] = 1
print(result)
```

## 5，向量化k-NN距离

```python
Q = 3
N = 5
D = 4
query = rng.normal(size=(Q, D))
database = rng.normal(size=(N, D))
print("query 的形状：", query.shape)
print("database 的形状：", database.shape)
query_expanded = query[:, None, :]          # (Q, 1, D)
database_expanded = database[None, :, :]    # (1, N, D)
print("扩展后 query 的形状：", query_expanded.shape)
print("扩展后 database 的形状：", database_expanded.shape)
# 利用广播计算每个查询向量和每个数据库向量的差
# (Q, 1, D) - (1, N, D) -> (Q, N, D)
differences = query_expanded - database_expanded
# 沿最后一个维度计算平方和
squared = (differences ** 2).sum(axis=-1)
# 开平方，得到欧氏距离矩阵
distances = np.sqrt(squared)
print("\n距离矩阵：")
print(distances)
print("\n距离矩阵形状：", distances.shape)
# 自检
assert query.shape == (Q, D)
assert database.shape == (N, D)
assert distances.shape == (Q, N)
assert np.all(distances >= 0)
```
